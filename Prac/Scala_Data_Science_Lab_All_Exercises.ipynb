# Exercise 13: Introduction to Arrays in Scala

## Problem Statement:
1. Create an array of integers with the values `2, 5, 9, 14, 20`.
2. Write a function that takes this array and returns the sum of its elements.
3. Next, create an array of strings with the names of five different fruits.
4. Write a function that concatenates all elements of this array, separated by commas.
5. Finally, iterate over the array of integers and print each element to the console.

---

## Theoretical Deep Dive: Arrays in Scala
- **Memory & Representation**:
  - In Scala, `Array[T]` is a fixed-size, mutable sequential collection.
  - Unlike purely functional collections, Scala arrays map directly onto **native JVM arrays** (e.g., `Array[Int]` becomes `int[]`, and `Array[String]` becomes `String[]`). This avoids object boxing overhead and achieves optimal cache-locality.
- **Rich Collection Integration via `ArrayOps`**:
  - The Scala compiler automatically wraps arrays with `ArrayOps` through implicit conversions. This gives JVM arrays rich functional methods such as `.sum`, `.foldLeft`, `.map`, `.filter`, and `.mkString`.
- **String Concatenation with `.mkString`**:
  - The `.mkString(delimiter)` method formats collections into a single delimited string efficiently using an internal `StringBuilder`.
- **Iteration with `.foreach`**:
  - `.foreach(f: T => Unit)` applies a side-effecting function (such as `println`) to every element sequentially.


```scala
// 1. Create an array of integers with values: 2, 5, 9, 14, 20
val intArray: Array[Int] = Array(2, 5, 9, 14, 20)

println(s"Created Integer Array: ${intArray.mkString("[", ", ", "]")}")
println(s"Array Length: ${intArray.length}")

// 2. Function that takes an integer array and returns the sum of its elements
def sumArray(arr: Array[Int]): Int = {
  // Method A: Built-in ArrayOps .sum method (highly optimized)
  // arr.sum

  // Method B: Functional accumulation using foldLeft
  arr.foldLeft(0)((acc, elem) => acc + elem)
}

val totalSum: Int = sumArray(intArray)
println(s"Sum of array elements: $totalSum")
```

```scala
// 3. Create an array of strings with the names of five different fruits
val fruitArray: Array[String] = Array("Apple", "Banana", "Cherry", "Mango", "Orange")

println(s"Created Fruits Array: ${fruitArray.mkString("[", ", ", "]")}")

// 4. Function that concatenates all elements of this array, separated by commas
def concatenateFruits(fruits: Array[String]): String = {
  // mkString joins all elements using the specified separator
  fruits.mkString(", ")
}

val concatenatedResult: String = concatenateFruits(fruitArray)
println(s"Concatenated Fruits String: $concatenatedResult")
```

```scala
// 5. Iterate over the array of integers and print each element to the console
println("--- Iterating with .foreach ---")
intArray.foreach { num =>
  println(s"Element: $num")
}

println("\n--- Alternative: Iterating with an indexed loop ---")
for (index <- intArray.indices) {
  println(s"Index $index -> Value: ${intArray(index)}")
}
```

# Exercise 14: Understanding Lists, Sets, and Tuples in Scala

## Problem Statement:
1. Create a `List` of the first five prime numbers.
2. Write a function that takes this list and returns a new list with each element squared.
3. Next, create a `Set` of unique characters from the string `"hello world"`.
4. Write a function that takes two sets and returns their intersection.
5. Finally, create a `Tuple` with three elements: an integer, a string, and a boolean; then access and print each element.

---

## Theoretical Deep Dive: Core Scala Collections
- **`List[T]`**:
  - Implemented as an immutable singly linked list made of a `head` (the first element) and a `tail` (the remainder of the list).
  - Prepending with `::` (cons) is $O(1)$ and purely functional because new lists share the tail structure without copying.
  - The `.map(f: A => B)` method returns a new list containing transformed values.
- **`Set[T]`**:
  - An immutable collection of unique elements (no duplicates).
  - Converting `"hello world".toSet` extracts the unique characters (`'h'`, `'e'`, `'l'`, `'o'`, `' '`, `'w'`, `'r'`, `'d'`).
  - Set intersection (`setA.intersect(setB)` or `setA & setB`) yields only elements present in both sets.
- **`Tuple` (Heterogeneous Container)**:
  - Unlike lists or arrays which require all elements to share the same type, tuples can group values of different types (e.g. `(Int, String, Boolean)`).
  - Positional accessors `_1`, `_2`, and `_3` are 1-indexed and statically type-checked by the compiler.


```scala
// 1. Create a list of the first five prime numbers
val primes: List[Int] = List(2, 3, 5, 7, 11)
println(s"Original Prime List: $primes")

// 2. Function that takes this list and returns a new list with each element squared
def squareList(lst: List[Int]): List[Int] = {
  // map applies the squaring lambda function to every element immutably
  lst.map(x => x * x)
}

val squaredPrimes = squareList(primes)
println(s"Squared Prime List:  $squaredPrimes")
```

```scala
// 3. Create a set of unique characters from the string "hello world"
val originalText = "hello world"
val uniqueChars: Set[Char] = originalText.toSet

println(s"Original text: '$originalText'")
println(s"Set of unique characters: $uniqueChars")
println(s"Notice duplicate letters like 'l' and 'o' appear only once!")

// 4. Function that takes two sets and returns their intersection
def setIntersection[T](setA: Set[T], setB: Set[T]): Set[T] = {
  // setA.intersect(setB) or setA & setB returns the common elements
  setA.intersect(setB)
}

val comparisonSet: Set[Char] = Set('e', 'l', 'o', 'z', '!')
val commonChars = setIntersection(uniqueChars, comparisonSet)

println(s"\nComparison Set: $comparisonSet")
println(s"Set Intersection: $commonChars")
```

```scala
// 5. Create a tuple with three elements: an integer, a string, and a boolean
val studentTuple: (Int, String, Boolean) = (101, "Data Science", true)
println(s"Complete Tuple: $studentTuple")

// Access and print each element using positional accessors _1, _2, _3
val id: Int             = studentTuple._1
val subject: String     = studentTuple._2
val isEnrolled: Boolean = studentTuple._3

println(s"\nElement 1 (_1 - Roll Number): $id")
println(s"Element 2 (_2 - Subject Name): $subject")
println(s"Element 3 (_3 - Is Enrolled):  $isEnrolled")

// Pattern matching unpacks all fields at once:
val (roll, sub, active) = studentTuple
println(s"Pattern Unpacked: ID=$roll, Subject=$sub, Active=$active")
```

# Exercise 15: Handling Collisions and Resolving Conflicts in HashMaps

## Problem Statement:
1. Create a `HashMap` with keys representing student names and values representing their grades, then insert multiple entries including a duplicate key with a different grade.
2. Write a function to merge two HashMaps, resolving conflicts by taking the **higher grade**.
3. Write a function to handle collisions by **chaining**, using `List`s to store multiple values for a single key.

---

## Theoretical Deep Dive: Hash Tables and Collision Resolution
- **Standard HashMap Semantics**:
  - In a standard key-value map, each key is uniquely associated with at most one value.
  - When you insert an entry with an existing key, the previous value is silently overwritten.
- **Conflict Resolution via Value Merging**:
  - When merging two datasets (such as exam 1 and exam 2 results), key clashes occur.
  - Rather than arbitrarily overwriting, we specify a reduction rule: `math.max(gradeA, gradeB)` to preserve the student's highest performance.
- **Collision Resolution via Separate Chaining**:
  - Separate chaining stores values within bucket collections (e.g. `Map[String, List[Int]]`).
  - When a duplicate key arrives (e.g. a student taking multiple quizzes), the new score is appended to the list rather than overwriting previous scores, preserving the full submission history.


```scala
import scala.collection.mutable

// 1. Create a mutable HashMap with student names and grades
val gradeMap = mutable.HashMap[String, Int]()
gradeMap.put("Alice", 85)
gradeMap.put("Bob", 78)
gradeMap.put("Charlie", 92)

println("--- Initial HashMap Entries ---")
gradeMap.foreach { case (name, grade) => println(s"  $name -> $grade") }

// Insert a duplicate key "Alice" with a different grade (e.g. 95)
println("\nInserting duplicate key 'Alice' with new grade 95...")
val oldGrade = gradeMap.put("Alice", 95)

println(s"Replaced previous grade: $oldGrade")
println("--- HashMap After Duplicate Key Insertion (Overwritten) ---")
gradeMap.foreach { case (name, grade) => println(s"  $name -> $grade") }
```

```scala
// 2. Function to merge two HashMaps, resolving conflicts by taking the HIGHER grade
def mergeResolvingHigherGrade(map1: Map[String, Int], map2: Map[String, Int]): Map[String, Int] = {
  val allKeys = map1.keySet ++ map2.keySet
  
  allKeys.map { key =>
    val resolvedGrade = (map1.get(key), map2.get(key)) match {
      case (Some(g1), Some(g2)) => math.max(g1, g2) // Both maps have key -> pick max!
      case (Some(g1), None)     => g1               // Only in map1
      case (None, Some(g2))     => g2               // Only in map2
      case _                    => 0
    }
    key -> resolvedGrade
  }.toMap
}

// Sample test data for two classes / exam sessions
val class1: Map[String, Int] = Map("Alice" -> 85, "Bob" -> 90, "David" -> 74)
val class2: Map[String, Int] = Map("Alice" -> 92, "Bob" -> 82, "Emma" -> 88)

println(s"Class 1 Grades: $class1")
println(s"Class 2 Grades: $class2")

val resolvedMap = mergeResolvingHigherGrade(class1, class2)
println(s"\nMerged Map (Conflicts resolved by higher grade): $resolvedMap")
```

```scala
// 3. Function to handle collisions by chaining using lists for multiple values
def insertWithChaining(
  map: Map[String, List[Int]],
  student: String,
  grade: Int
): Map[String, List[Int]] = {
  val existingGrades = map.getOrElse(student, List.empty[Int])
  // Append the new grade to the student's grade list
  map + (student -> (existingGrades :+ grade))
}

// Build chained map from a sequence of student submissions containing duplicates
def buildChainedMap(submissions: Seq[(String, Int)]): Map[String, List[Int]] = {
  submissions.foldLeft(Map.empty[String, List[Int]]) { (accMap, submission) =>
    insertWithChaining(accMap, submission._1, submission._2)
  }
}

// Sequence containing multiple submissions for Alice and Bob
val allSubmissions = Seq(
  ("Alice", 85),
  ("Bob", 78),
  ("Charlie", 92),
  ("Alice", 95),  // Alice Quiz 2
  ("Bob", 88),    // Bob Quiz 2
  ("Alice", 90),  // Alice Quiz 3
  ("David", 81)
)

val chainedResults = buildChainedMap(allSubmissions)

println("--- Chained Hash Table Results (All Historical Grades Preserved) ---")
chainedResults.foreach { case (student, grades) =>
  val average = grades.sum.toDouble / grades.length
  println(f"Student: $student%-8s | All Scores: ${grades.mkString("[", ", ", "]")}%-18s | Best: ${grades.max}%2d | Avg: $average%.2f")
}
```

# Exercise 16: Exploring Advanced Functional Combinators Beyond the Basic Set

## Problem Statement:
1. Create a list of integers from `1` to `10`.
2. Use `filter` to create a new list with only even numbers.
3. Use `map` to create a new list where each element is multiplied by `3`.
4. Use `flatMap` to create a list of tuples where each integer is paired with its square.
5. Finally, use `foldLeft` to calculate the product of all elements in the list.

---

## Theoretical Deep Dive: Functional Combinators
- **Declarative vs. Imperative**:
  - Instead of managing loop counter variables and mutating state, combinators express **what** transformation should occur through pure, side-effect-free functions.
- **The Combinators in Detail**:
  1. **`filter(p: A => Boolean): List[A]`**:
     - Retains only elements for which the predicate `p` evaluates to `true`.
  2. **`map(f: A => B): List[B]`**:
     - Applies a 1-to-1 transformation `f` across each item in the collection.
  3. **`flatMap(f: A => IterableOnce[B]): List[B]`**:
     - Maps each element to a collection and flattens all inner collections into one sequential list.
     - By returning `List((x, x * x))` for each integer, `flatMap` yields a flat list of `(Int, Int)` pairs.
  4. **`foldLeft(initial: B)(op: (B, A) => B): B`**:
     - Traverses from left to right, combining elements using an accumulator.
     - With initial accumulator `1L`, cumulatively multiplying yields the factorial $10! = 3,628,800$.


```scala
// 1. Create a list of integers from 1 to 10
val numbers: List[Int] = (1 to 10).toList
println(s"Original List: $numbers")

// 2. Use filter to create a new list with only even numbers
val evenNumbers: List[Int] = numbers.filter(x => x % 2 == 0)
println(s"\n1. filter (Even numbers only): $evenNumbers")

// 3. Use map to create a new list where each element is multiplied by 3
val multipliedBy3: List[Int] = numbers.map(x => x * 3)
println(s"2. map (Each element multiplied by 3): $multipliedBy3")

// 4. Use flatMap to create a list of tuples where each integer is paired with its square
val pairedWithSquare: List[(Int, Int)] = numbers.flatMap { x =>
  List((x, x * x))
}
println(s"3. flatMap (Each integer paired with its square):")
println(pairedWithSquare)

// 5. Use foldLeft to calculate the product of all elements in the list (10!)
val product: Long = numbers.foldLeft(1L) { (accumulator, currentVal) =>
  accumulator * currentVal
}
println(s"\n4. foldLeft (Product of all elements 1*2*...*10): $product")
```

### Summary of Combinator Transformations

| Combinator | Lambda Type | Input Sample | Output | Purpose |
| :--- | :--- | :--- | :--- | :--- |
| **`filter`** | `x => x % 2 == 0` | `1 to 10` | `List(2, 4, 6, 8, 10)` | Select only even integers |
| **`map`** | `x => x * 3` | `1 to 10` | `List(3, 6, 9, ..., 30)` | Scale each integer by 3 |
| **`flatMap`** | `x => List((x, x*x))` | `1 to 10` | `List((1,1), (2,4), ..., (10,100))` | Pair each item with its square |
| **`foldLeft`** | `(acc, x) => acc * x` | `1 to 10` | `3628800` | Compute product of list |


# Exercise 17: Case Classes in Scala

## Problem Statement:
1. Create a case class `Person` with fields `name`, `age`, and `city`.
2. Create a list of `Person` objects representing five different individuals.
3. Write a function to filter out people older than 30.
4. Next, write a function that groups people by their `city`.
5. Then, write a function that transforms each `Person` object into a string in the format:
   `"Name is Age years old and lives in City"`.
6. Finally, write a function to sort the list of people by `age` in ascending order.

---

## Theoretical Deep Dive: Case Classes in Scala
- **What makes Case Classes special?**
  - Case classes are immutable data-modeling constructs tailored for functional programming and domain-driven design.
  - The Scala compiler automatically generates:
    1. **Public immutable fields**: Every constructor parameter is implicitly a `val`.
    2. **Companion factory method**: Instances are instantiated cleanly as `Person("Alice", 25, "NYC")` without the `new` keyword.
    3. **Structural equality**: `.equals` and `.hashCode` compare field values rather than JVM memory addresses.
    4. **Readable string representation**: Generates an informative `.toString` automatically.
    5. **Pattern matching extractor**: Generates `unapply` for destructuring in `match` statements.


```scala
// 1. Define the Person case class
case class Person(name: String, age: Int, city: String)

// 2. Create a list of Person objects representing five different individuals
val people: List[Person] = List(
  Person("Alice", 25, "New York"),
  Person("Bob", 35, "San Francisco"),
  Person("Charlie", 28, "New York"),
  Person("Diana", 42, "London"),
  Person("Edward", 30, "London")
)

println("--- Initial People List ---")
people.foreach(println)

// 3. Function to filter out people older than 30 (keeps people with age <= 30)
def filterOutOlderThan30(list: List[Person]): List[Person] = {
  list.filter(_.age <= 30)
}

val youngerOrEqual30 = filterOutOlderThan30(people)
println("\n1. Filtered (Excluding people older than 30, keeping age <= 30):")
youngerOrEqual30.foreach(println)

// 4. Function that groups people by their city
def groupByCity(list: List[Person]): Map[String, List[Person]] = {
  list.groupBy(_.city)
}

val peopleByCity = groupByCity(people)
println("\n2. Grouped by City:")
peopleByCity.foreach { case (city, group) =>
  val names = group.map(_.name).mkString(", ")
  println(s"  $city -> [$names]")
}

// 5. Function to transform each Person into "Name is Age years old and lives in City"
def formatPeople(list: List[Person]): List[String] = {
  list.map(p => s"${p.name} is ${p.age} years old and lives in ${p.city}")
}

val formattedPeople = formatPeople(people)
println("\n3. Formatted Person Descriptions:")
formattedPeople.foreach(println)

// 6. Function to sort the list of people by age in ascending order
def sortByAge(list: List[Person]): List[Person] = {
  list.sortBy(_.age)
}

val sortedByAge = sortByAge(people)
println("\n4. Sorted by Age (Ascending):")
sortedByAge.foreach(p => println(f"  ${p.name}%-8s: ${p.age} years old (${p.city})"))
```

# Exercise 18: Data Manipulation with Saddle & Interactive Visualization with Vegas

## Problem Statement:
1. Use **Saddle** to load and manipulate a dataset of fruit prices and quantities over time.
2. Filter the dataset for **apples** and calculate their **average price**.
3. Visualize the price trend over time using **Vegas**.

---

## Theoretical Deep Dive: Saddle & Vegas
- **Saddle (`org.saddle`)**:
  - Saddle is the high-performance data library for Scala, bringing the indexed, vector-backed architecture of Python's **Pandas** to the JVM.
  - Core structures:
    - `Vec[T]`: Primitive array wrapper enabling vectorized calculations and missing-data handling.
    - `Series[K, V]`: Indexed 1D vector.
    - `Frame[RowKey, ColKey, Val]`: 2D indexed tabular matrix.
- **Vegas (`vegas._`)**:
  - Declarative statistical visualization library wrapping **Vega-Lite**.
  - Maps data dimensions to visual encoding channels (`.encodeX`, `.encodeY`) and marks (`Line`, `Bar`, `Point`).


```scala
// Ivy imports for Almond kernel in Colab
import $ivy.`org.saddle::saddle-core:1.3.4`
import $ivy.`org.vegas-viz::vegas:0.3.11`
import $ivy.`org.vegas-viz::vegas-spark:0.3.11`

println("Saddle and Vegas libraries successfully imported!")
```

```scala
import org.saddle._

// 1. Create vectors for Date, Fruit, Price, and Quantity
val dates      = Vec("2023-10-01", "2023-10-02", "2023-10-03", "2023-10-04", "2023-10-05", "2023-10-01", "2023-10-02")
val fruitNames = Vec("Apple", "Apple", "Apple", "Apple", "Apple", "Banana", "Orange")
val prices     = Vec(2.40, 2.65, 2.50, 2.90, 3.10, 1.20, 1.80)
val quantities = Vec(120.0, 110.0, 140.0, 95.0, 130.0, 250.0, 180.0)

// Construct a Saddle 2D Frame
val fruitFrame: Frame[Int, String, Any] = Frame(
  "Date"     -> dates,
  "Fruit"    -> fruitNames,
  "Price"    -> prices,
  "Quantity" -> quantities
)

println("--- Full Fruit Dataset (Saddle Frame) ---")
println(fruitFrame)
```

```scala
// 2. Filter for apples and calculate their average price
val fruitVec = fruitFrame.col("Fruit").toVec
val isAppleMask = fruitVec.map(_ == "Apple")

// Filter rows where fruit is Apple
val appleFrame = fruitFrame.filterRows(r => isAppleMask.at(r))
println("--- Filtered Dataset: Apples Only ---")
println(appleFrame)

// Extract Apple prices as Double Vec and compute mean
val applePrices = appleFrame.col("Price").mapValues(_.asInstanceOf[Double]).toVec
val avgApplePrice = applePrices.mean

println(f"Number of Apple Observations: ${applePrices.length}")
println(f"Average Apple Price: $$${avgApplePrice}%.2f")
```

```scala
import vegas._
import vegas.render.HTML5._

// 3. Prepare data and visualize price trend over time using Vegas
val applePlotData = Seq(
  Map("Date" -> "2023-10-01", "Price" -> 2.40),
  Map("Date" -> "2023-10-02", "Price" -> 2.65),
  Map("Date" -> "2023-10-03", "Price" -> 2.50),
  Map("Date" -> "2023-10-04", "Price" -> 2.90),
  Map("Date" -> "2023-10-05", "Price" -> 3.10)
)

val plot = Vegas("Apple Price Trend Over Time")
  .withData(applePlotData)
  .encodeX("Date", Nom)
  .encodeY("Price", Quant)
  .mark(Line)

println("--- Generated Vega-Lite Visualization Specification ---")
println(plot.asVega_Lite)

// In Jupyter/Almond, calling plot.show renders the interactive visualization inline!
```

# Exercise 19: Numerical Computations with Breeze & Dynamic Visualizations with Plotly.Scala

## Problem Statement:
1. Use the **Breeze** library to calculate total revenue for each fruit type by multiplying price and quantity from a dataset (`Price * Quantity`).
2. Create an interactive bar chart to visualize these revenues using **Plotly.Scala**.

---

## Theoretical Deep Dive: Breeze & Plotly.Scala
- **Breeze (`breeze.linalg._`)**:
  - Breeze is the fundamental scientific and numerical computing library for Scala, equivalent to Python's **NumPy**.
  - Provides SIMD-accelerated dense vectors (`DenseVector`) and matrices (`DenseMatrix`).
  - Supports element-wise vector operations: `*:*` (element-wise multiplication), `+:+`, and aggregations like `sum()`.
- **Plotly.Scala (`plotly._`)**:
  - Type-safe Scala DSL for **Plotly.js**.
  - Renders rich interactive charts (tooltips, zoom, pan, hover) directly in Jupyter/Colab notebooks or exports them to standalone HTML files.


```scala
// Ivy imports for Almond kernel in Colab
import $ivy.`org.scalanlp::breeze:2.1.0`
import $ivy.`org.plotly-scala::plotly-render:0.8.4`

println("Breeze and Plotly.Scala libraries successfully imported!")
```

```scala
import breeze.linalg._

// 1. Define fruit types and numerical vectors
val fruitNames = Seq("Apple", "Banana", "Cherry", "Mango", "Orange")

// Unit prices in USD
val prices: DenseVector[Double] = DenseVector(2.50, 1.20, 4.00, 3.50, 2.00)

// Quantities sold
val quantities: DenseVector[Double] = DenseVector(150.0, 300.0, 80.0, 110.0, 200.0)

println("--- Price Vector ---")
println(prices)

println("\n--- Quantity Vector ---")
println(quantities)

// 2. Calculate Total Revenue using Breeze element-wise multiplication: Price * Quantity
val totalRevenues: DenseVector[Double] = prices *:* quantities

println("\n--- Total Revenue Calculated with Breeze ---")
for (i <- fruitNames.indices) {
  println(f"${fruitNames(i)}%-8s : $$${prices(i)}%.2f x ${quantities(i)}%.0f units = $$${totalRevenues(i)}%.2f")
}

val grandTotal = sum(totalRevenues)
println(f"\nTotal Overall Store Revenue: $$${grandTotal}%.2f")
```

```scala
import plotly._
import plotly.element._
import plotly.layout._
import plotly.Plotly._

// 3. Prepare data and create an interactive Bar chart using Plotly.Scala
val xFruits: Seq[String] = fruitNames
val yRevenues: Seq[Double] = totalRevenues.toArray.toSeq

val barTrace = Bar(
  x = xFruits,
  y = yRevenues,
  name = "Revenue ($)"
)

val chartLayout = Layout(
  title = "Total Fruit Revenue by Fruit Type (Breeze + Plotly.Scala)",
  xaxis = Axis(title = "Fruit Name"),
  yaxis = Axis(title = "Total Revenue (USD $)"),
  bargap = 0.25
)

// Export to standalone interactive HTML chart
Plotly.plot(
  path = "fruit_revenues_chart.html",
  traces = Seq(barTrace),
  layout = chartLayout
)

println("Interactive Plotly Bar Chart successfully generated and exported to 'fruit_revenues_chart.html'!")
// In Jupyter/Almond, barTrace.plot(layout = chartLayout) renders the interactive chart directly inline!
```
