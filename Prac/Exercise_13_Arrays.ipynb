# Exercise 13: Introduction to Arrays in Scala

## Problem Statement:
1. Create an array of integers with the values `2, 5, 9, 14, 20`.
2. Write a function that takes this array and returns the sum of its elements.
3. Next, create an array of strings with the names of five different fruits.
4. Write a function that concatenates all elements of this array, separated by commas.
5. Finally, iterate over the array of integers and print each element to the console.

---

## Theoretical Deep Dive: Arrays in Scala
- **Memory & Representation**:
  - In Scala, `Array[T]` is a fixed-size, mutable sequential collection.
  - Unlike purely functional collections, Scala arrays map directly onto **native JVM arrays** (e.g., `Array[Int]` becomes `int[]`, and `Array[String]` becomes `String[]`). This avoids object boxing overhead and achieves optimal cache-locality.
- **Rich Collection Integration via `ArrayOps`**:
  - The Scala compiler automatically wraps arrays with `ArrayOps` through implicit conversions. This gives JVM arrays rich functional methods such as `.sum`, `.foldLeft`, `.map`, `.filter`, and `.mkString`.
- **String Concatenation with `.mkString`**:
  - The `.mkString(delimiter)` method formats collections into a single delimited string efficiently using an internal `StringBuilder`.
- **Iteration with `.foreach`**:
  - `.foreach(f: T => Unit)` applies a side-effecting function (such as `println`) to every element sequentially.


```scala
// 1. Create an array of integers with values: 2, 5, 9, 14, 20
val intArray: Array[Int] = Array(2, 5, 9, 14, 20)

println(s"Created Integer Array: ${intArray.mkString("[", ", ", "]")}")
println(s"Array Length: ${intArray.length}")

// 2. Function that takes an integer array and returns the sum of its elements
def sumArray(arr: Array[Int]): Int = {
  // Method A: Built-in ArrayOps .sum method (highly optimized)
  // arr.sum

  // Method B: Functional accumulation using foldLeft
  arr.foldLeft(0)((acc, elem) => acc + elem)
}

val totalSum: Int = sumArray(intArray)
println(s"Sum of array elements: $totalSum")
```

```scala
// 3. Create an array of strings with the names of five different fruits
val fruitArray: Array[String] = Array("Apple", "Banana", "Cherry", "Mango", "Orange")

println(s"Created Fruits Array: ${fruitArray.mkString("[", ", ", "]")}")

// 4. Function that concatenates all elements of this array, separated by commas
def concatenateFruits(fruits: Array[String]): String = {
  // mkString joins all elements using the specified separator
  fruits.mkString(", ")
}

val concatenatedResult: String = concatenateFruits(fruitArray)
println(s"Concatenated Fruits String: $concatenatedResult")
```

```scala
// 5. Iterate over the array of integers and print each element to the console
println("--- Iterating with .foreach ---")
intArray.foreach { num =>
  println(s"Element: $num")
}

println("\n--- Alternative: Iterating with an indexed loop ---")
for (index <- intArray.indices) {
  println(s"Index $index -> Value: ${intArray(index)}")
}
```
