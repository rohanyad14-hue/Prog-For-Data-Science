# Exercise 19: Numerical Computations with Breeze & Dynamic Visualizations with Plotly.Scala

## Problem Statement:
1. Use the **Breeze** library to calculate total revenue for each fruit type by multiplying price and quantity from a dataset (`Price * Quantity`).
2. Create an interactive bar chart to visualize these revenues using **Plotly.Scala**.

---

## Theoretical Deep Dive: Breeze & Plotly.Scala
- **Breeze (`breeze.linalg._`)**:
  - Breeze is the fundamental scientific and numerical computing library for Scala, equivalent to Python's **NumPy**.
  - Provides SIMD-accelerated dense vectors (`DenseVector`) and matrices (`DenseMatrix`).
  - Supports element-wise vector operations: `*:*` (element-wise multiplication), `+:+`, and aggregations like `sum()`.
- **Plotly.Scala (`plotly._`)**:
  - Type-safe Scala DSL for **Plotly.js**.
  - Renders rich interactive charts (tooltips, zoom, pan, hover) directly in Jupyter/Colab notebooks or exports them to standalone HTML files.


```scala
// Ivy imports for Almond kernel in Colab
import $ivy.`org.scalanlp::breeze:2.1.0`
import $ivy.`org.plotly-scala::plotly-render:0.8.4`

println("Breeze and Plotly.Scala libraries successfully imported!")
```

```scala
import breeze.linalg._

// 1. Define fruit types and numerical vectors
val fruitNames = Seq("Apple", "Banana", "Cherry", "Mango", "Orange")

// Unit prices in USD
val prices: DenseVector[Double] = DenseVector(2.50, 1.20, 4.00, 3.50, 2.00)

// Quantities sold
val quantities: DenseVector[Double] = DenseVector(150.0, 300.0, 80.0, 110.0, 200.0)

println("--- Price Vector ---")
println(prices)

println("\n--- Quantity Vector ---")
println(quantities)

// 2. Calculate Total Revenue using Breeze element-wise multiplication: Price * Quantity
val totalRevenues: DenseVector[Double] = prices *:* quantities

println("\n--- Total Revenue Calculated with Breeze ---")
for (i <- fruitNames.indices) {
  println(f"${fruitNames(i)}%-8s : $$${prices(i)}%.2f x ${quantities(i)}%.0f units = $$${totalRevenues(i)}%.2f")
}

val grandTotal = sum(totalRevenues)
println(f"\nTotal Overall Store Revenue: $$${grandTotal}%.2f")
```

```scala
import plotly._
import plotly.element._
import plotly.layout._
import plotly.Plotly._

// 3. Prepare data and create an interactive Bar chart using Plotly.Scala
val xFruits: Seq[String] = fruitNames
val yRevenues: Seq[Double] = totalRevenues.toArray.toSeq

val barTrace = Bar(
  x = xFruits,
  y = yRevenues,
  name = "Revenue ($)"
)

val chartLayout = Layout(
  title = "Total Fruit Revenue by Fruit Type (Breeze + Plotly.Scala)",
  xaxis = Axis(title = "Fruit Name"),
  yaxis = Axis(title = "Total Revenue (USD $)"),
  bargap = 0.25
)

// Export to standalone interactive HTML chart
Plotly.plot(
  path = "fruit_revenues_chart.html",
  traces = Seq(barTrace),
  layout = chartLayout
)

println("Interactive Plotly Bar Chart successfully generated and exported to 'fruit_revenues_chart.html'!")
// In Jupyter/Almond, barTrace.plot(layout = chartLayout) renders the interactive chart directly inline!
```
